# Base Emulator for any statistics

In principle, you can use `Emulator` class to load any trained files to directly predict the corresponding statistics. The `Emulator` class is designed to be flexible and can handle various types of statistical models, making it a versatile tool for statistical analysis.

```python
from jet.emulator import Emulator

emu = Emulator.load("pklin.gp.npz")
print(type(emu).__name__)
print("input :", emu.x_spec.names)
print("output:", emu.y_spec.name, "->", emu.y_spec.n_bins, "values")
print("extra_arrays:", emu.extra_arrays.keys())
```
